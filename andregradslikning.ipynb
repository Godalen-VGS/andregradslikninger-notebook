## Test løsningen din

Cellen under kjører funksjonen din mot 17 faste og 30 tilfeldige likninger og sammenlikner med fasit. Det er de samme tilfellene som brukes i den automatiske rettingen, så kjør den gjerne underveis for å få rask tilbakemelding.

Har likningen to løsninger, spiller det ingen rolle hvilken rekkefølge du returnerer dem i.

````python
import math
import random

INGEN = "Likningen har ingen løsning!"
TO = [(1, -5, 6), (1, 0, -4), (2, -3, -5), (-1, 2, 3), (1, 1, -1), (3, -2, -7), (1, -3, 0)]
EN = [(1, -4, 4), (2, 4, 2), (-3, 6, -3), (1, 0, 0), (4, -4, 1), (9, 6, 1)]
INGEN_TILFELLER = [(1, 2, 5), (-2, 1, -3), (1, 0, 1), (5, -2, 1)]

def _naer_avrundingsgrense(x):
    return abs((abs(x) * 100) % 1 - 0.5) < 1e-6

def _tilfeldige(antall, seed=2025):
    rng = random.Random(seed)
    tilfeller = []
    while len(tilfeller) < antall:
        a = rng.randint(-100, 100) / 10
        b = rng.randint(-100, 100) / 10
        c = rng.randint(-100, 100) / 10
        d = b**2 - 4*a*c
        if a == 0 or abs(d) < 1e-6:
            continue
        if d > 0 and any(_naer_avrundingsgrense((-b + s * math.sqrt(d)) / (2*a)) for s in (1, -1)):
            continue
        tilfeller.append((a, b, c))
    return tilfeller

def _fasit(a, b, c):
    d = b**2 - 4*a*c
    if d > 0:
        return (round((-b + math.sqrt(d)) / (2*a), 2), round((-b - math.sqrt(d)) / (2*a), 2))
    if d == 0:
        return round(-b / (2*a), 2)
    return INGEN

def _stemmer(svar, forventet):
    if isinstance(forventet, tuple):
        return isinstance(svar, (tuple, list)) and len(svar) == 2 and sorted(svar) == sorted(forventet)
    return svar == forventet

def test_losninger():
    tilfeller = TO + EN + INGEN_TILFELLER + _tilfeldige(30)
    feil = 0
    for a, b, c in tilfeller:
        fasit = _fasit(a, b, c)
        try:
            svar = losninger(a, b, c)
        except Exception as e:
            svar = f"krasjet med {type(e).__name__}: {e}"
        if not _stemmer(svar, fasit):
            feil += 1
            print(f"FEIL for a={a}, b={b}, c={c}: fikk {svar!r}, forventet {fasit!r}")
            if feil >= 5:
                print("... stopper etter 5 feil.")
                break
    if feil == 0:
        print(f"Alle {len(tilfeller)} testene bestått!")

test_losninger()
````
