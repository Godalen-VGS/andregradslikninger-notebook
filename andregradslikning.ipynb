Cellen under kjører funksjonen din mot 200 tilfeldig genererte likninger og sammenlikner med fasit. Dette er den samme testen som brukes til automatisk retting (se README i denne mappen) — kjør den gjerne underveis for å få rask tilbakemelding.


````python
import random
import math

def _generer_tilfeldige_koeffisienter():
    a = random.randint(-100, 100) / 10
    if round(a, 2) == 0:
        a += 1
    b = random.randint(-100, 100) / 10
    c = random.randint(-100, 100) / 10
    if random.randint(1, 20) == 1:
        c = b**2 / (4 * a)
    return a, b, c

def test_losninger(antall=200):
    feil = 0
    for _ in range(antall):
        a, b, c = _generer_tilfeldige_koeffisienter()
        diskriminant = b**2 - 4*a*c
        if diskriminant > 0:
            fasit = (round((-b + math.sqrt(diskriminant)) / (2*a), 2),
                     round((-b - math.sqrt(diskriminant)) / (2*a), 2))
        elif diskriminant == 0:
            fasit = round(-b / (2*a), 2)
        else:
            fasit = "Likningen har ingen løsning!"

        svar = losninger(a, b, c)
        if svar != fasit:
            feil += 1
            print(f"FEIL for a={a}, b={b}, c={c}: fikk {svar!r}, forventet {fasit!r}")
            if feil >= 5:
                print("... stopper etter 5 feil.")
                break
    if feil == 0:
        print(f"Alle {antall} testene bestått!")

test_losninger()
````
